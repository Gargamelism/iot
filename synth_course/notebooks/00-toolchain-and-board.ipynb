```python
import sys, numpy, scipy, matplotlib
import numpy as np
import matplotlib.pyplot as plt
from synthlib import FS, play, write_wav, read_wav, note_name, check, ROOT

print("python    ", sys.version.split()[0])
print("numpy     ", numpy.__version__)
print("scipy     ", scipy.__version__)
print("matplotlib", matplotlib.__version__)
print("notebooks ", ROOT)
```

## A tone you should hear

Two seconds of A4 (440 Hz). Press play — keep your volume low the first time.

```python
t = np.arange(2 * FS) / FS
tone = 0.3 * np.sin(2 * np.pi * 440 * t)
play(tone, normalize=False)
plt.figure(figsize=(9, 2.5))
plt.plot(t[:400] * 1000, tone[:400])
plt.xlabel("ms"); plt.title("first 9 ms of the tone"); plt.show()
```

## Files and helpers work

```python
import tempfile, pathlib
path = write_wav(pathlib.Path(tempfile.gettempdir()) / "setup-check.wav", tone)
back, fs = read_wav(path)
check(fs == FS and len(back) == len(tone), "WAV round-trip keeps rate and length")
check(np.max(np.abs(back - tone)) < 1e-4, "WAV round-trip keeps the samples (16-bit)")
check(note_name(440) == "A4" and note_name(261.6) == "C4", "note names match the firmware's notes.h")
```
